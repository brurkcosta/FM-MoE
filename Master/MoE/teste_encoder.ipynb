```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("encoder_features.csv")
```

```python
df
```

```python
df_corr = df.select_dtypes(include='number').corr().abs()
plt.figure(figsize=(30,30))
sns.heatmap(df_corr, annot=True, fmt='.2f')
```

```python
c = df_corr.values.copy()
np.fill_diagonal(c, np.nan)         
print("correlação média entre features:", round(np.nanmean(c), 3))
```
